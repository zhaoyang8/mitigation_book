## Distorted density for climate models <a href="https://github.com/korito1416/two-capital-climate-change/blob/new_model_boundary/python/ZeroShockTrajectories_simulate.py" class="code-link" target="_blank" rel="noopener"></a>


The decision-maker has a baseline probability $dP_t(\theta)$ for each time instant, $t$, and makes an adjustment for ambiguity by solving

$$
\begin{aligned}
\min_{q,\;\int_\Theta q(\theta)\,dP_t(\theta)=1}\;
      &\;\chi \int_\Theta q(\theta)\,\log q(\theta)\,dP_t(\theta) \\
\text{s.t. }\;&
      \int_\Theta \mu\!\bigl[X_t,\Phi^{\ast}(X_t)\bigr]\,
               q(\theta)\,dP_t(\theta)=\bar h(X_t),
\end{aligned}
$$

where $\Phi^{\ast}$ is the robustly optimal solution and $\bar h$ is deduced from a preference specification with misspecification aversion.
The minimizing density takes the form of

$$
q_t^{\ast}(\theta)
\;\propto\;
\exp\!\Bigl[-\zeta\,\mu\!\bigl(X_t,a^{\ast}(X_t)\bigr)\Bigr],
\qquad
\text{with }\zeta\text{ chosen so the constraint holds}.
$$

## Distorted density for the time of the first jump  <a href="https://github.com/korito1416/two-capital-climate-change/blob/new_model_boundary/python/FeymannKacs_simulate_NewPlug.py" class="code-link" target="_blank" rel="noopener"></a>

By conditioning first on the state-variable history, we find that the probability that a jump has occurred by time $t$ is  

$$
F_t \;=\; 1 - \exp\!\Bigl[ - \sum_{\ell=1}^{L} \int_{0}^{t} 
        \mathcal J^{\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr],
$$

which increases in $t$.  
The implied density obtained by differentiating with respect to $t$ is  

$$
f_t \;=\; \sum_{\ell=1}^{L} \mathcal J^{\ell}(X_t)\,
        \exp\!\Bigl[ - \int_{0}^{t} \mathcal J^{\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr].
$$

We express $f$ as  

$$
f_t \;=\; \sum_{\ell=1}^{L} f^{\ell}(t),
$$

where  

$$
f_t^{\ell} \overset{\mathrm{def}}{=} \mathcal J^{\ell}(X_t)\,
\exp\!\Bigl[ - \sum_{\ell=1}^{L} \int_{0}^{t} 
            \mathcal J^{\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr].
$$
