## Distorted density for climate models <a href="https://github.com/korito1416/two-capital-climate-change/blob/damage_tech_jump_model/python/ZeroShockTrajectories_simulate.py" class="code-link" target="_blank" rel="noopener"></a>


The decision-maker has a baseline probability $dP_t(\theta)$ that weights each of the 144 climate models equally at every time instant, $t$, and makes an adjustment for ambiguity by solving Problem 3.2 of the paper:

$$
\min_{Q_t \ge 0,\; \int_\Theta Q_t(\theta)\,dP_t(\theta)=1}\;
\frac{\partial V}{\partial y}(X_t)\,\mathcal{E}_t \int_\Theta \theta\,Q_t(\theta)\,dP_t(\theta)
\;+\;
\xi_a \int_\Theta Q_t(\theta)\,\log Q_t(\theta)\,dP_t(\theta)
$$

with the minimizing relative density given by an exponential tilt:

$$
Q_t^{\ast}(\theta)
\;\propto\;
\exp\!\Bigl[-\frac{1}{\xi_a}\,\frac{\partial V}{\partial y}(X_t)\,\mathcal{E}_t\,\theta\Bigr].
$$

## Distorted density for the time of the first jump  <a href="https://github.com/korito1416/two-capital-climate-change/blob/damage_tech_jump_model/python/FeymannKacs_simulate_NewPlug.py" class="code-link" target="_blank" rel="noopener"></a>

By conditioning first on the state-variable history, we find that the probability that a jump has occurred by time $t$ under the uncertainty-adjusted probabilities is

$$
F_t \;=\; 1 - \exp\!\Bigl[ - \int_{0}^{t} \sum_{\ell=1}^{L}
        G^{*\ell}_\tau\,\mathcal J^{\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr],
$$

which increases in $t$.
The implied density obtained by differentiating with respect to $t$ is

$$
f_t \;=\; \sum_{\ell=1}^{L} G^{*\ell}_t\,\mathcal J^{\ell}(X_t)\,
        \exp\!\Bigl[ - \int_{0}^{t} \sum_{\hat\ell=1}^{L}
        G^{*\hat\ell}_\tau\,\mathcal J^{\hat\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr].
$$

We unpack this density by jump type,

$$
f_t \;=\; \sum_{\ell=1}^{L} f_t^{\ell},
$$

where

$$
f_t^{\ell} \overset{\mathrm{def}}{=} G^{*\ell}_t\,\mathcal J^{\ell}(X_t)\,
\exp\!\Bigl[ - \int_{0}^{t} \sum_{\hat\ell=1}^{L}
            G^{*\hat\ell}_\tau\,\mathcal J^{\hat\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr],
\qquad \ell = 1,\dots,L.
$$
